valid sudoku

```python
import numpy as np
s = np.array([[5,3,4,6,7,8,9,1,2],
              [6,7,2,1,9,5,3,4,8],                   # SUDOKU MEI KYA HOTA HAI HAR INDIVIDUAL ROW KO PLUS KARKE 45 AATA HAI AND HAR INDIVIDUAL COLUMN KO PLUS KARKE BHI 45 AATA HAI
              [1,9,8,3,4,2,5,6,7],

              [8,5,9,7,6,1,4,2,3],
              [4,2,6,8,5,3,7,9,1],
              [7,1,3,9,2,4,8,5,6],

              [9,6,1,5,3,7,2,8,4],
              [2,8,7,4,1,9,6,3,5],
              [3,4,5,2,8,6,1,7,9]])
```

```python
b = np.sum(s,axis=1)
b==45 # as sari rows ka sum is 45 hence my sudoku is valid
```

```python
np.sum(s,axis=0) # for colmns it is valid
```

```python
for i in range(0,9,3):
    for j in range(0,9,3):
        a = s[i:i+3,j:j+3]
        # print(a)                       # our 3 by 3 blocks perfectly print ho rhe hai
        # now if u want to check their sum
        print(np.sum(a))    # as all 3 by 3 ka block ka bhi sum is 45 therefore our sudoku is valid
```
